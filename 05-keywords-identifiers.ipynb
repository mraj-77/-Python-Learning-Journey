# Keywords

**Python is case sensitive**

- *Keywords*: reserved words, special meanings
- Cannot use as **variable names**

```python
# Python has 35 keywords.
import keyword
print(keyword.kwlist)
l = len(keyword.kwlist)
print("Total number of keywords in Python:", l)
```

# Identifiers

**Identifier** are name for var, func, class, module, etc.

**Rules**:
- Start with letter or `_`
- Followed by letters, digits, `_`
- Not a Python keyword

```python
name = 'Saurabh'
print(name)
```

```python
_ = 'Saurabh'
print(_)
```

```python
@ = 'Saurabh'
print(@)  # @ is not a valid variable name in Python. Attempting to assign a value to @ will result in a SyntaxError.
```

```python
2 = 'Saurabh'
print(2)  # 2 is not a valid variable name in Python. Variable names cannot start with a digit. Attempting to assign a value to 2 will result in a SyntaxError.
```

```python
first_name = 'Saurabh'
print(first_name)
```

```python
False = 'Saurabh'
print(False)  # False is a keyword in Python, and it cannot be used as a variable name. Attempting to assign a value to False will result in a SyntaxError.
```
