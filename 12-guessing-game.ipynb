```python
jackpot = random.randint(1, 100)
guess = int(input("Guess the no. : "))
counter = 1
while guess != jackpot:
    if guess < jackpot:
        print("Guess higher")
    else:
        print("Guess lower")   
    guess = int(input("Guess the no. : "))
    counter += 1
print("Congratulations! You guessed the correct number", jackpot)
print("You took", counter, "attempts")
```
